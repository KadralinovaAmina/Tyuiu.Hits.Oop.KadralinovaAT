```
public class Product
{
    private string name;
    private decimal price;
    private int quantity;
    protected string category;
    internal string description;

    public string Name
    {
        get { return name; }
        set { name = value; }
    }
    public decimal Price
    {
        get { return price; }
        set { price = value; }
    }
    public int Quantity
    {
        get { return quantity; }
        set { quantity = value; }
    }
    public string Category
    {
        get { return category; }
        set { category = value; }
    }
    public string Description
    {
        get { return description; }
        set { description = value; }
    }

    public Product(string name, decimal price, int quantity, string category, string description)
    {
        this.name = name;
        this.price = price;
        this.quantity = quantity;
        this.category = category;
        this.description = description;
    }

    public decimal GetTotalPrice()
    {
        return price * quantity;
    }

    public void UpdateQuantity(int amount)
    {
        quantity = amount;
    }

    public string GetProductInfo()
    {
        return $"Название: {name}, цена: {price}, количество: {quantity}, категория: {category}, описание: {description}";
    }
}

public class Seller
{
    private string name;
    private string employeeId;
    protected decimal salary;
    internal string contactInfo;
    private List<Product> products = new List<Product>();

    public string Name
    {
        get { return name; }
        set { name = value; }
    }
    public string EmployeeId
    {
        get { return employeeId; }
        set { employeeId = value; }
    }
    public decimal Salary
    {
        get { return salary; }
        set { salary = value; }
    }
    public string ContactInfo
    {
        get { return contactInfo; }
        set { contactInfo = value; }
    }
    public List<Product> Products
    {
        get { return products; }
    }

    public Seller(string name, string employeeId, decimal salary, string contactInfo)
    {
        this.name = name;
        this.employeeId = employeeId;
        this.salary = salary;
        this.contactInfo = contactInfo;
    }

    public void AddProduct(Product product)
    {
        products.Add(product);
        Console.WriteLine($"Товар '{product.Name}' добавлен в список товаров продавца {name}.");
    }

    public void SellProduct(Product product, int quantity)
    {
        if (product.Quantity >= quantity)
        {
            product.UpdateQuantity(product.Quantity - quantity);
            Console.WriteLine($"Товар '{product.Name}' продан в количестве {quantity} шт. Остаток: {product.Quantity}");
        }
        else
        {
            Console.WriteLine($"Недостаточно товара '{product.Name}' для продажи. В наличии: {product.Quantity} шт.");
        }
    }
}

public class Store
{
    private string storeName;
    private string location;
    public string storeHours;
    private List<Seller> sellers = new List<Seller>();
    private List<Product> products = new List<Product>();

    public string StoreName
    {
        get { return storeName; }
        set { storeName = value; }
    }
    public string Location
    {
        get { return location; }
        set { location = value; }
    }
    public string StoreHours
    {
        get { return storeHours; }
        set { storeHours = value; }
    }
    public List<Seller> Sellers
    {
        get { return sellers; }
    }
    public List<Product> Products
    {
        get { return products; }
    }

    public Store(string storeName, string location, string storeHours)
    {
        this.storeName = storeName;
        this.location = location;
        this.storeHours = storeHours;
    }

    public void AddSeller(Seller seller)
    {
        sellers.Add(seller);
        Console.WriteLine($"Продавец {seller.Name} добавлен в магазин {storeName}.");
    }

    public void AddProductToStore(Product product)
    {
        products.Add(product);
        Console.WriteLine($"Товар '{product.Name}' добавлен в магазин {storeName}.");
    }

    public void ListProducts()
    {
        Console.WriteLine($"\n--- Товары в магазине {storeName} ---");
        bool hasProducts = false;
        foreach (var product in products)
        {
            Console.WriteLine(product.GetProductInfo());
            hasProducts = true;
        }
        if (!hasProducts)
        {
            Console.WriteLine("Товаров в магазине нет.");
        }
    }

    public string GetStoreInfo()
    {
        return $"Магазин: {storeName}, Адрес: {location}, График работы: {storeHours}";
    }
}

Console.WriteLine("--- Создание магазина ---\n");
var myStore = new Store("Мвидео", "г. Тюмень, ул. Мельникайте, д. 152", "10:00 - 22:00");
Console.WriteLine(myStore.GetStoreInfo());

var seller1 = new Seller("Samsung", "IPP052", 8000m, "samsung@gmail.com");
myStore.AddSeller(seller1);

var product1 = new Product("Ноутбук", 55000m, 20, "Электроника", "Ноутбук с подсветкой клавиатуры");
var product2 = new Product("Телевизор", 75000m, 30, "Электроника", "Телевизор со Смарт ТВ");
var product3 = new Product("Смартфон", 15000m, 30, "Электроника", "Смартфон складной");
var product4 = new Product("Микроволновка", 35000m, 20, "Электроника", "Микроволновка с мультиповаром");

seller1.AddProduct(product1);
seller1.AddProduct(product2);
seller1.AddProduct(product3);
seller1.AddProduct(product4);

myStore.AddProductToStore(product1);
myStore.AddProductToStore(product2);
myStore.AddProductToStore(product3);
myStore.AddProductToStore(product4);

Console.WriteLine("\n--- Список товаров ---");
myStore.ListProducts();

Console.WriteLine($"Общая стоимость ноутбуков на складе: {product1.GetTotalPrice():C}");

seller1.SellProduct(product1, 5);

Console.WriteLine($"Общая стоимость ноутбуков после продажи: {product1.GetTotalPrice():C}");

product2.UpdateQuantity(30);

Console.WriteLine("\n--- Обновленный список товаров ---");
myStore.ListProducts();
```
