<h3 style="color:DodgerBlue">Описание задачи:</h3>

Создать базовый класс Inventory в C#, который будет представлять информацию о
наличии товаров на складе. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.

<h3 style="color:DodgerBlue">Требования к базовому классу Inventory:</h3>

* Атрибуты: ID склада (WarehouseId), Название склада (WarehouseName),
Общий объем хранения (StorageCapacity).
* Методы:
    * GetStorageStatus(): метод для получения статуса доступного
    пространства на складе. 
    * AddItem(Item item): метод для добавления товара на склад.
    * RemoveItem(Item item): метод для удаления товара со склада.
    
<h3 style="color:DodgerBlue">Требования к производным классам:</h3>

1. ПерсональныйСклад (PersonalInventory): Должен содержать
дополнительные атрибуты, такие как Владелец склада (OwnerName).
Метод GetStorageStatus() должен быть переопределен для отображения
информации о владельце склада вместе с статусом хранения.

2. ГрупповойСклад (GroupInventory): Должен содержать дополнительные
атрибуты, такие как Группа товаров (ProductGroup). Метод AddItem() должен
быть переопределен для добавления информации о группе товаров при
добавлении нового товара.

3. АвтоматизированныйСклад (AutomatedInventory) (если требуется третий
класс): Должен содержать дополнительные атрибуты, такие как
Автоматизация уровня (AutomationLevel). Метод RemoveItem() должен быть
переопределен для добавления информации о уровне автоматизации при
удалении товара.

<h3 style="color:DodgerBlue">Дополнительное задание:</h3>

* Добавьте к сущестующим классам конструктора классов с использованием гетторов и сетторов и реализуйте взаимодействие объектов между собой;

* Добавьте к сущестующим классам (базовыму и производным 3-4 атрибута и метода) и реализуйте простое, сложное и множественное наследование.

```
public interface ITrackable
{
    void Track();
}

public interface IReportable
{
    void GenerateReport();
}

public interface ISecurable
{
    void Lock();
}
public class Item
{
    private string _name;
    private int _id;
    private double _weight;
    private string _category;

    public string Name
    {
        get { return _name; }
        set { _name = value; }
    }

    public int Id
    {
        get { return _id; }
        set { _id = value; }
    }

    public double Weight
    {
        get { return _weight; }
        set { _weight = value; }
    }

    public string Category
    {
        get { return _category; }
        set { _category = value; }
    }

    public Item(string name, int id, double weight, string category)
    {
        Name = name;
        Id = id;
        Weight = weight;
        Category = category;
    }

    public void ShowDetails()
    {
        Console.WriteLine($"Товар: {Name}, ID: {Id}, Вес: {Weight} кг, Категория: {Category}");
    }

    public void UpdateWeight(double newWeight)
    {
        Weight = newWeight;
        Console.WriteLine($"Вес товара '{Name}' обновлен: {Weight} кг");
    }
}
public class Inventory
{
    private int _warehouseId;
    private string _warehouseName;
    private double _storageCapacity;
    private string _location;
    private int _itemCount;

    public int WarehouseId
    {
        get { return _warehouseId; }
        set { _warehouseId = value; }
    }

    public string WarehouseName
    {
        get { return _warehouseName; }
        set { _warehouseName = value; }
    }

    public double StorageCapacity
    {
        get { return _storageCapacity; }
        set { _storageCapacity = value; }
    }

    public string Location
    {
        get { return _location; }
        set { _location = value; }
    }

    public int ItemCount
    {
        get { return _itemCount; }
        set { _itemCount = value; }
    }

    public Inventory(int id, string name, double capacity, string location)
    {
        WarehouseId = id;
        WarehouseName = name;
        StorageCapacity = capacity;
        Location = location;
        ItemCount = 0;
    }

    public virtual void GetStorageStatus()
    {
        Console.WriteLine($"'{WarehouseName}', ID: {WarehouseId}");
        Console.WriteLine($"Локация: {Location}");
        Console.WriteLine($"Общая вместимость: {StorageCapacity} ед.");
        Console.WriteLine($"Товаров на складе: {ItemCount}");
        Console.WriteLine("Статус: Пространство доступно.");
    }

    public virtual void AddItem(Item item)
    {
        ItemCount++;
        Console.WriteLine($"Товар '{item.Name}' добавлен на '{WarehouseName}'");
    }

    public virtual void RemoveItem(Item item)
    {
        ItemCount--;
        Console.WriteLine($"Товар '{item.Name}' удален со склада '{WarehouseName}'");
    }

    public void TransferItem(Item item, Inventory target)
    {
        Console.WriteLine($"Перемещение товара '{item.Name}' с '{WarehouseName}' на '{target.WarehouseName}'");
        this.RemoveItem(item);
        target.AddItem(item);
    }

    public void ShowLocation()
    {
        Console.WriteLine($"Склад '{WarehouseName}' находится в {Location}");
    }
}
public class PersonalInventory : Inventory, ITrackable
{
    private string _ownerName;
    private string _ownerPhone;
    private bool _isPrivate;

    public string OwnerName
    {
        get { return _ownerName; }
        set { _ownerName = value; }
    }

    public string OwnerPhone
    {
        get { return _ownerPhone; }
        set { _ownerPhone = value; }
    }

    public bool IsPrivate
    {
        get { return _isPrivate; }
        set { _isPrivate = value; }
    }

    public PersonalInventory(int id, string name, double capacity, string location,
                             string owner, string phone, bool isPrivate)
        : base(id, name, capacity, location)
    {
        OwnerName = owner;
        OwnerPhone = phone;
        IsPrivate = isPrivate;
    }

    public override void GetStorageStatus()
    {
        base.GetStorageStatus();
        Console.WriteLine($"Владелец склада: {OwnerName}");
        Console.WriteLine($"Телефон: {OwnerPhone}");
        Console.WriteLine($"Приватный: {IsPrivate}");
    }

    public void GiveItemTo(Item item, Inventory other)
    {
        Console.WriteLine($"{OwnerName} передает товар '{item.Name}' на склад '{other.WarehouseName}'");
        this.RemoveItem(item);
        other.AddItem(item);
    }

    public void Track()
    {
        Console.WriteLine($"Отслеживание персонального склада '{WarehouseName}' владельца {OwnerName}");
    }

    public void ShowOwnerInfo()
    {
        Console.WriteLine($"Владелец: {OwnerName}, Телефон: {OwnerPhone}");
    }
}
public class GroupInventory : Inventory, IReportable
{
    private string _productGroup;
    private int _groupSize;
    private string _responsiblePerson;

    public string ProductGroup
    {
        get { return _productGroup; }
        set { _productGroup = value; }
    }

    public int GroupSize
    {
        get { return _groupSize; }
        set { _groupSize = value; }
    }

    public string ResponsiblePerson
    {
        get { return _responsiblePerson; }
        set { _responsiblePerson = value; }
    }

    public GroupInventory(int id, string name, double capacity, string location,
                          string group, int groupSize, string responsible)
        : base(id, name, capacity, location)
    {
        ProductGroup = group;
        GroupSize = groupSize;
        ResponsiblePerson = responsible;
    }

    public override void AddItem(Item item)
    {
        ItemCount++;
        Console.WriteLine($"Товар '{item.Name}' добавлен в группу товаров '{ProductGroup}' на '{WarehouseName}'");
    }

    public void ReceiveItemFrom(Item item, Inventory other)
    {
        Console.WriteLine($"Склад '{WarehouseName}' принимает товар '{item.Name}' от '{other.WarehouseName}'");
        other.RemoveItem(item);
        this.AddItem(item);
    }

    public void GenerateReport()
    {
        Console.WriteLine($"Отчет по группе '{ProductGroup}': {GroupSize} товаров, ответственный: {ResponsiblePerson}");
    }
}
public class AutomatedInventory : Inventory, ISecurable
{
    private int _automationLevel;
    private string _robotModel;
    private bool _isLocked;

    public int AutomationLevel
    {
        get { return _automationLevel; }
        set { _automationLevel = value; }
    }

    public string RobotModel
    {
        get { return _robotModel; }
        set { _robotModel = value; }
    }

    public bool IsLocked
    {
        get { return _isLocked; }
        set { _isLocked = value; }
    }

    public AutomatedInventory(int id, string name, double capacity, string location,
                              int level, string robotModel)
        : base(id, name, capacity, location)
    {
        AutomationLevel = level;
        RobotModel = robotModel;
        IsLocked = false;
    }

    public override void RemoveItem(Item item)
    {
        ItemCount--;
        Console.WriteLine($"Робот '{RobotModel}' (уровень {AutomationLevel}) удалил товар '{item.Name}' со склада '{WarehouseName}'");
    }

    public void AutoTransfer(Item item, Inventory target)
    {
        Console.WriteLine($"Робот '{RobotModel}' перемещает товар '{item.Name}' на '{target.WarehouseName}'");
        this.RemoveItem(item);
        target.AddItem(item);
    }

    public void Lock()
    {
        IsLocked = true;
        Console.WriteLine($"Автоматизированный склад '{WarehouseName}' заблокирован");
    }

    public void Unlock()
    {
        IsLocked = false;
        Console.WriteLine($"Автоматизированный склад '{WarehouseName}' разблокирован");
    }
}
public class SmartPersonalInventory : PersonalInventory
{
    private string _smartSystemVersion;
    private bool _voiceControl;

    public string SmartSystemVersion
    {
        get { return _smartSystemVersion; }
        set { _smartSystemVersion = value; }
    }

    public bool VoiceControl
    {
        get { return _voiceControl; }
        set { _voiceControl = value; }
    }

    public SmartPersonalInventory(int id, string name, double capacity, string location,
                                  string owner, string phone, bool isPrivate,
                                  string version, bool voiceControl)
        : base(id, name, capacity, location, owner, phone, isPrivate)
    {
        SmartSystemVersion = version;
        VoiceControl = voiceControl;
    }

    public override void GetStorageStatus()
    {
        base.GetStorageStatus();
        Console.WriteLine($"Умная система: версия {SmartSystemVersion}");
        Console.WriteLine($"Голосовое управление: {VoiceControl}");
    }

    public void ActivateVoiceControl()
    {
        VoiceControl = true;
        Console.WriteLine($"Голосовое управление активировано на '{WarehouseName}'");
    }
}
public class HybridInventory : Inventory, ITrackable, IReportable, ISecurable
{
    private string _hybridType;

    public string HybridType
    {
        get { return _hybridType; }
        set { _hybridType = value; }
    }

    public HybridInventory(int id, string name, double capacity, string location, string hybridType)
        : base(id, name, capacity, location)
    {
        HybridType = hybridType;
    }

    public void Track()
    {
        Console.WriteLine($"Отслеживание гибридного склада '{WarehouseName}'");
    }

    public void GenerateReport()
    {
        Console.WriteLine($"Отчет по гибридному складу '{WarehouseName}' типа '{HybridType}'");
    }

    public void Lock()
    {
        Console.WriteLine($"Гибридный склад '{WarehouseName}' заблокирован");
    }

    public override void GetStorageStatus()
    {
        base.GetStorageStatus();
        Console.WriteLine($"Тип гибридного склада: {HybridType}");
    }
}

Item item1 = new Item("Телевизор", 101, 15.5, "Электроника");
Item item2 = new Item("Смартфон", 102, 0.3, "Электроника");
Item item3 = new Item("Книги", 103, 5.0, "Литература");
Item item4 = new Item("Тетради", 104, 1.2, "Канцелярия");
Item item5 = new Item("Доски", 105, 20.0, "Стройматериалы");
Item item6 = new Item("Брус", 106, 25.0, "Стройматериалы");

Console.WriteLine("-------- Простое наследование --------\n");

Console.WriteLine("--- Персональный склад ---");
PersonalInventory pers = new PersonalInventory(172, "Кладовой склад", 300.0, "Тюмень",
                                                "Амина Кадралинова", "+7-952-123-45-67", true);
pers.GetStorageStatus();
pers.AddItem(item3);
pers.RemoveItem(item4);
pers.Track();
pers.ShowOwnerInfo();
Console.WriteLine();

Console.WriteLine("--- Групповой склад ---");
GroupInventory group = new GroupInventory(212, "Склад электроники", 500.0, "Тюменьг",
                                           "Электроника", 50, "Иван Стариков");
group.GetStorageStatus();
group.AddItem(item1);
group.RemoveItem(item2);
group.GenerateReport();
Console.WriteLine();

Console.WriteLine("--- Автоматизированный склад ---");
AutomatedInventory auto = new AutomatedInventory(385, "Строительный склад", 800.0, "Тюмень",
                                                  3, "R6-D7");
auto.GetStorageStatus();
auto.AddItem(item5);
auto.RemoveItem(item6);
auto.Lock();
auto.Unlock();
Console.WriteLine();

Console.WriteLine("-------- Сложное наследование --------\n");

Console.WriteLine("--- Умный персональный склад ---");
SmartPersonalInventory smart = new SmartPersonalInventory(500, "Умный склад", 400.0, "Тюмень",
                                                           "Петр Смирнов", "+7-952-987-65-52", true,
                                                           "v2.0", false);
smart.GetStorageStatus();
smart.ActivateVoiceControl();
Console.WriteLine();

Console.WriteLine("-------- Множественное наследование --------\n");

Console.WriteLine("--- Гибридный склад ---");
HybridInventory hybrid = new HybridInventory(700, "Гибридный склад", 1000.0, "Тюмень", "Умный+Автоматизированный");
hybrid.GetStorageStatus();
hybrid.Track();
hybrid.GenerateReport();
hybrid.Lock();
Console.WriteLine();

Console.WriteLine("-------- Взаимодействие объектов --------\n");

Console.WriteLine("--- Передача товара ---");
pers.GiveItemTo(item3, group);
Console.WriteLine();

Console.WriteLine("--- Прием товара ---");
group.ReceiveItemFrom(item5, auto);
Console.WriteLine();

Console.WriteLine("--- Автоматическое перемещение ---");
auto.AutoTransfer(item6, pers);
Console.WriteLine();

Console.WriteLine("--- Универсальное перемещение ---");
group.TransferItem(item1, pers);
Console.WriteLine();
```
