```
using System;
using System.Collections.Generic;
using System.Linq;
public interface IInstallable
{
    void Install();
}
public abstract class Component
{
    public string Name { get; set; }
    public double Price { get; set; }

    public Component(string name, double price)
    {
        Name = name;
        Price = price;
    }
    public virtual void GetInfo()
    {
        Console.WriteLine($"Компонент: {Name}, Цена: {Price} руб.");
    }
    public override string ToString()
    {
        return $"[Компонент: {Name}, Цена: {Price}]";
    }
    public static bool operator ==(Component a, Component b)
    {
        if (ReferenceEquals(a, b)) return true;
        if (a is null || b is null) return false;
        return a.Name == b.Name && a.Price == b.Price;
    }

    public static bool operator !=(Component a, Component b)
    {
        return !(a == b);
    }

    public override bool Equals(object obj)
    {
        return obj is Component c && this == c;
    }

    public override int GetHashCode()
    {
        return HashCode.Combine(Name, Price);
    }
}
public class CPU : Component, IInstallable
{
    public int Cores { get; set; }

    public CPU(string name, double price, int cores) : base(name, price)
    {
        Cores = cores;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Тип: Процессор, Ядра: {Cores}");
    }

    public void Install()
    {
        Console.WriteLine($"Процессор '{Name}' установлен в сокет.");
    }
}

public class RAM : Component, IInstallable
{
    public int SizeGB { get; set; }

    public RAM(string name, double price, int sizeGB) : base(name, price)
    {
        SizeGB = sizeGB;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Тип: Оперативная память, Объем: {SizeGB} ГБ");
    }

    public void Install()
    {
        Console.WriteLine($"Оперативная память '{Name}' установлена в слот.");
    }
}

public class GPU : Component, IInstallable
{
    public int VideoMemoryGB { get; set; }

    public GPU(string name, double price, int videoMemoryGB) : base(name, price)
    {
        VideoMemoryGB = videoMemoryGB;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Тип: Видеокарта, Видеопамять: {VideoMemoryGB} ГБ");
    }

    public void Install()
    {
        Console.WriteLine($"Видеокарта '{Name}' установлена в слот PCI-E.");
    }
}

public class HDD : Component, IInstallable
{
    public int CapacityTB { get; set; }

    public HDD(string name, double price, int capacityTB) : base(name, price)
    {
        CapacityTB = capacityTB;
    }

    public override void GetInfo()
    {
        base.GetInfo();
        Console.WriteLine($"Тип: Жесткий диск, Объем: {CapacityTB} ТБ");
    }

    public void Install()
    {
        Console.WriteLine($"Жесткий диск '{Name}' установлен в отсек.");
    }
}
public class Computer
{
    public string Model { get; set; }
    private List<Component> _components = new List<Component>();

    public Computer(string model)
    {
        Model = model;
    }
    public void AddComponent<T>(T component) where T : Component, IInstallable
    {
        _components.Add(component);
        component.Install();
        Console.WriteLine($"Компонент '{component.Name}' добавлен в компьютер {Model}.");
    }
    public void RemoveComponent<T>(T component) where T : Component
    {
        if (_components.Contains(component))
        {
            _components.Remove(component);
            Console.WriteLine($"Компонент '{component.Name}' удален из компьютера {Model}.");
        }
        else
        {
            Console.WriteLine($"Компонент '{component.Name}' не найден в компьютере {Model}.");
        }
    }
    public double GetTotalPrice()
    {
        return _components.Sum(c => c.Price);
    }

    public void ShowAllComponents()
    {
        Console.WriteLine($"\n--- Состав компьютера {Model} ---");
        foreach (var comp in _components)
        {
            comp.GetInfo();
            Console.WriteLine("---");
        }
    }
}

public static class ComputerFactory
{
    public static Computer CreateGamingPC<T1, T2, T3>(T1 cpu, T2 ram, T3 gpu) 
        where T1 : Component, IInstallable
        where T2 : Component, IInstallable
        where T3 : Component, IInstallable
    {
        Computer pc = new Computer("Gaming PC");
        pc.AddComponent(cpu);
        pc.AddComponent(ram);
        pc.AddComponent(gpu);
        return pc;
    }

    public static Computer CreateOfficePC<T1, T2>(T1 cpu, T2 ram)
        where T1 : Component, IInstallable
        where T2 : Component, IInstallable
    {
        Computer pc = new Computer("Office PC");
        pc.AddComponent(cpu);
        pc.AddComponent(ram);
        return pc;
    }
}

CPU cpu1 = new CPU("Intel Core i9", 50000, 16);
RAM ram1 = new RAM("Kingston DDR5", 15000, 32);
GPU gpu1 = new GPU("NVIDIA RTX 4090", 200000, 24);
HDD hdd1 = new HDD("Seagate Barracuda", 8000, 4);

Console.WriteLine("----------- ИНФОРМАЦИЯ О КОМПОНЕНТАХ -----------");
cpu1.GetInfo();
Console.WriteLine();
ram1.GetInfo();
Console.WriteLine();
gpu1.GetInfo();

Console.WriteLine("\n----------- ПЕРЕГРУЗКА ToString() -----------");
Console.WriteLine($"Строковое представление CPU: {cpu1.ToString()}");

Console.WriteLine("\n----------- ПЕРЕГРУЗКА ОПЕРАТОРОВ СРАВНЕНИЯ -----------");
CPU cpu2 = new CPU("Intel Core i9", 50000, 16);
CPU cpu3 = new CPU("AMD Ryzen 9", 45000, 12);

Console.WriteLine($"cpu1 == cpu2: {cpu1 == cpu2}");
Console.WriteLine($"cpu1 == cpu3: {cpu1 == cpu3}");
Console.WriteLine($"cpu1 != cpu3: {cpu1 != cpu3}");

Console.WriteLine("\n----------- РАБОТА С КЛАССОМ COMPUTER -----------");
Computer myPC = new Computer("Мой ПК");
myPC.AddComponent(cpu1);
myPC.AddComponent(ram1);
myPC.AddComponent(gpu1);
myPC.AddComponent(hdd1);

myPC.ShowAllComponents();

Console.WriteLine($"\nОбщая стоимость компьютера: {myPC.GetTotalPrice()} руб.");

Console.WriteLine("\n--- Удаление компонента ---");
myPC.RemoveComponent(hdd1);
Console.WriteLine($"Новая стоимость после удаления HDD: {myPC.GetTotalPrice()} руб.");

Console.WriteLine("\n----------- ФАБРИКА -----------");
CPU factoryCpu = new CPU("AMD Ryzen 5", 25000, 6);
RAM factoryRam = new RAM("Corsair DDR4", 8000, 16);
GPU factoryGpu = new GPU("AMD Radeon RX 6600", 40000, 8);

Computer gamingPC = ComputerFactory.CreateGamingPC(factoryCpu, factoryRam, factoryGpu);
Console.WriteLine($"\nСтоимость игрового ПК из фабрики: {gamingPC.GetTotalPrice()} руб.");

Computer officePC = ComputerFactory.CreateOfficePC(new CPU("Intel Core i3", 10000, 4), new RAM("Patriot DDR4", 4000, 8));
Console.WriteLine($"Стоимость офисного ПК из фабрики: {officePC.GetTotalPrice()} руб.");
```
