<h3 style="color:DodgerBlue">Описание задачи:</h3>

Создать базовый класс Inventory в C#, который будет представлять информацию о
наличии товаров на складе. На основе этого класса разработать 2-3 производных
класса, демонстрирующих принципы наследования и полиморфизма. В каждом из
классов должны быть реализованы новые атрибуты и методы, а также
переопределены некоторые методы базового класса для демонстрации
полиморфизма.

<h3 style="color:DodgerBlue">Требования к базовому классу Inventory:</h3>

* Атрибуты: ID склада (WarehouseId), Название склада (WarehouseName),
Общий объем хранения (StorageCapacity).
* Методы:
    * GetStorageStatus(): метод для получения статуса доступного
    пространства на складе. 
    * AddItem(Item item): метод для добавления товара на склад.
    * RemoveItem(Item item): метод для удаления товара со склада.
    
<h3 style="color:DodgerBlue">Требования к производным классам:</h3>

1. ПерсональныйСклад (PersonalInventory): Должен содержать
дополнительные атрибуты, такие как Владелец склада (OwnerName).
Метод GetStorageStatus() должен быть переопределен для отображения
информации о владельце склада вместе с статусом хранения.

2. ГрупповойСклад (GroupInventory): Должен содержать дополнительные
атрибуты, такие как Группа товаров (ProductGroup). Метод AddItem() должен
быть переопределен для добавления информации о группе товаров при
добавлении нового товара.

3. АвтоматизированныйСклад (AutomatedInventory) (если требуется третий
класс): Должен содержать дополнительные атрибуты, такие как
Автоматизация уровня (AutomationLevel). Метод RemoveItem() должен быть
переопределен для добавления информации о уровне автоматизации при
удалении товара.

```
using System;
using System.Collections.Generic;

public interface ITrackable { void Track(); }
public interface IReportable { void GenerateReport(); }
public interface ISecurable { void Lock(); }

public class Item
{
    private string _name;
    private int _id;
    private double _weight;
    private string _category;
    private string _supplier;
    private DateTime _expiryDate;

    public string Name { get { return _name; } set { _name = value; } }
    public int Id { get { return _id; } set { _id = value; } }
    public double Weight { get { return _weight; } set { _weight = value; } }
    public string Category { get { return _category; } set { _category = value; } }
    public string Supplier { get { return _supplier; } set { _supplier = value; } }
    public DateTime ExpiryDate { get { return _expiryDate; } set { _expiryDate = value; } }

    public Item(string name, int id, double weight, string category, string supplier, DateTime expiryDate)
    {
        Name = name;
        Id = id;
        Weight = weight;
        Category = category;
        Supplier = supplier;
        ExpiryDate = expiryDate;
    }

    public void ShowDetails()
    {
        Console.WriteLine($"Товар: {Name}, ID: {Id}, Вес: {Weight} кг, Категория: {Category}, Поставщик: {Supplier}, Годен до: {ExpiryDate:dd.MM.yyyy}");
    }

    public void UpdateWeight(double newWeight)
    {
        Weight = newWeight;
        Console.WriteLine($"Вес товара '{Name}' обновлен: {Weight} кг");
    }
}

public class Inventory
{
    private int _warehouseId;
    private string _warehouseName;
    private double _storageCapacity;
    private string _location;
    private int _itemCount;
    private string _manager;
    private string _securityCode;

    public int WarehouseId { get { return _warehouseId; } set { _warehouseId = value; } }
    public string WarehouseName { get { return _warehouseName; } set { _warehouseName = value; } }
    public double StorageCapacity { get { return _storageCapacity; } set { _storageCapacity = value; } }
    public string Location { get { return _location; } set { _location = value; } }
    public int ItemCount { get { return _itemCount; } set { _itemCount = value; } }
    public string Manager { get { return _manager; } set { _manager = value; } }
    public string SecurityCode { get { return _securityCode; } set { _securityCode = value; } }

    public Inventory(int id, string name, double capacity, string location, string manager, string securityCode)
    {
        WarehouseId = id;
        WarehouseName = name;
        StorageCapacity = capacity;
        Location = location;
        ItemCount = 0;
        Manager = manager;
        SecurityCode = securityCode;
    }

    public virtual void GetStorageStatus()
    {
        Console.WriteLine($"'{WarehouseName}', ID: {WarehouseId}");
        Console.WriteLine($"Локация: {Location}, Менеджер: {Manager}, Код безопасности: {SecurityCode}");
        Console.WriteLine($"Вместимость: {StorageCapacity} ед., Товаров: {ItemCount}");
    }

    public virtual void Audit()
    {
        Console.WriteLine($"Аудит склада '{WarehouseName}' пройден успешно.");
    }

    public virtual void AddItem(Item item)
    {
        ItemCount++;
        Console.WriteLine($"Товар '{item.Name}' добавлен на '{WarehouseName}'");
    }

    public virtual void RemoveItem(Item item)
    {
        ItemCount--;
        Console.WriteLine($"Товар '{item.Name}' удален со склада '{WarehouseName}'");
    }

    public void AddItem(string itemName)
    {
        ItemCount++;
        Console.WriteLine($"Товар '{itemName}' добавлен на '{WarehouseName}'");
    }

    public void TransferItem(Item item, Inventory target)
    {
        Console.WriteLine($"Перемещение товара '{item.Name}' с '{WarehouseName}' на '{target.WarehouseName}'");
        this.RemoveItem(item);
        target.AddItem(item);
    }
}

public class PersonalInventory : Inventory, ITrackable
{
    private string _ownerName;
    private string _ownerPhone;
    private bool _isPrivate;

    public string OwnerName { get { return _ownerName; } set { _ownerName = value; } }
    public string OwnerPhone { get { return _ownerPhone; } set { _ownerPhone = value; } }
    public bool IsPrivate { get { return _isPrivate; } set { _isPrivate = value; } }

    public PersonalInventory(int id, string name, double capacity, string location, string manager, string securityCode,
                             string owner, string phone, bool isPrivate)
        : base(id, name, capacity, location, manager, securityCode)
    {
        OwnerName = owner;
        OwnerPhone = phone;
        IsPrivate = isPrivate;
    }

    public override void GetStorageStatus()
    {
        base.GetStorageStatus();
        Console.WriteLine($"Владелец: {OwnerName}, Телефон: {OwnerPhone}, Приватный: {IsPrivate}");
    }

    public override void Audit()
    {
        base.Audit();
        Console.WriteLine($"Проверен персональный склад владельца {OwnerName}");
    }

    public void ChangeOwner(string newOwner)
    {
        OwnerName = newOwner;
        Console.WriteLine($"Новый владелец: {OwnerName}");
    }

    public void ChangeOwner(string newOwner, string newPhone)
    {
        OwnerName = newOwner;
        OwnerPhone = newPhone;
        Console.WriteLine($"Новый владелец: {OwnerName}, телефон: {OwnerPhone}");
    }

    public void GiveItemTo(Item item, Inventory other)
    {
        Console.WriteLine($"{OwnerName} передает товар '{item.Name}' на склад '{other.WarehouseName}'");
        this.RemoveItem(item);
        other.AddItem(item);
    }

    public void Track()
    {
        Console.WriteLine($"Отслеживание персонального склада '{WarehouseName}'");
    }
}

public class GroupInventory : Inventory, IReportable
{
    private string _productGroup;
    private int _groupSize;
    private string _responsiblePerson;

    public string ProductGroup { get { return _productGroup; } set { _productGroup = value; } }
    public int GroupSize { get { return _groupSize; } set { _groupSize = value; } }
    public string ResponsiblePerson { get { return _responsiblePerson; } set { _responsiblePerson = value; } }

    public GroupInventory(int id, string name, double capacity, string location, string manager, string securityCode,
                          string group, int groupSize, string responsible)
        : base(id, name, capacity, location, manager, securityCode)
    {
        ProductGroup = group;
        GroupSize = groupSize;
        ResponsiblePerson = responsible;
    }

    public override void AddItem(Item item)
    {
        ItemCount++;
        Console.WriteLine($"Товар '{item.Name}' добавлен в группу '{ProductGroup}' на '{WarehouseName}'");
    }

    public void AddToGroup(string itemName)
    {
        ItemCount++;
        Console.WriteLine($"Товар '{itemName}' добавлен в группу '{ProductGroup}' (по имени)");
    }

    public void GenerateReport()
    {
        Console.WriteLine($"Отчет по группе '{ProductGroup}': ответственный {ResponsiblePerson}");
    }
}

public class AutomatedInventory : Inventory, ISecurable
{
    private int _automationLevel;
    private string _robotModel;
    private bool _isLocked;
    private int _batteryLevel;
    private DateTime _maintenanceDate;

    public int AutomationLevel { get { return _automationLevel; } set { _automationLevel = value; } }
    public string RobotModel { get { return _robotModel; } set { _robotModel = value; } }
    public bool IsLocked { get { return _isLocked; } set { _isLocked = value; } }
    public int BatteryLevel { get { return _batteryLevel; } set { _batteryLevel = value; } }
    public DateTime MaintenanceDate { get { return _maintenanceDate; } set { _maintenanceDate = value; } }

    public AutomatedInventory(int id, string name, double capacity, string location, string manager, string securityCode,
                              int level, string robotModel, int batteryLevel, DateTime maintenanceDate)
        : base(id, name, capacity, location, manager, securityCode)
    {
        AutomationLevel = level;
        RobotModel = robotModel;
        IsLocked = false;
        BatteryLevel = batteryLevel;
        MaintenanceDate = maintenanceDate;
    }

    public override void RemoveItem(Item item)
    {
        ItemCount--;
        Console.WriteLine($"Робот '{RobotModel}' удалил товар '{item.Name}' со склада '{WarehouseName}'");
    }

    public void Charge()
    {
        BatteryLevel = 100;
        Console.WriteLine($"Робот заряжен на 100%");
    }

    public void Charge(int amount)
    {
        BatteryLevel += amount;
        if (BatteryLevel > 100) BatteryLevel = 100;
        Console.WriteLine($"Робот заряжен на {BatteryLevel}%");
    }

    public void Lock()
    {
        IsLocked = true;
        Console.WriteLine($"Автоматизированный склад '{WarehouseName}' заблокирован");
    }
}

public class SmartPersonalInventory : PersonalInventory
{
    private string _smartSystemVersion;
    private bool _voiceControl;
    private string _aiAssistantName;

    public string SmartSystemVersion { get { return _smartSystemVersion; } set { _smartSystemVersion = value; } }
    public bool VoiceControl { get { return _voiceControl; } set { _voiceControl = value; } }
    public string AiAssistantName { get { return _aiAssistantName; } set { _aiAssistantName = value; } }

    public SmartPersonalInventory(int id, string name, double capacity, string location, string manager, string securityCode,
                                  string owner, string phone, bool isPrivate,
                                  string version, bool voiceControl, string aiName)
        : base(id, name, capacity, location, manager, securityCode, owner, phone, isPrivate)
    {
        SmartSystemVersion = version;
        VoiceControl = voiceControl;
        AiAssistantName = aiName;
    }

    public override void GetStorageStatus()
    {
        base.GetStorageStatus();
        Console.WriteLine($"Умная система: {SmartSystemVersion}, AI: {AiAssistantName}");
    }

    public void SetAssistant(string name)
    {
        AiAssistantName = name;
        Console.WriteLine($"AI-ассистент: {AiAssistantName}");
    }

    public void SetAssistant(string name, string version)
    {
        AiAssistantName = name;
        SmartSystemVersion = version;
        Console.WriteLine($"AI-ассистент: {AiAssistantName}, версия: {SmartSystemVersion}");
    }
}

public class HybridInventory : Inventory, ITrackable, IReportable, ISecurable
{
    private string _hybridType;

    public string HybridType { get { return _hybridType; } set { _hybridType = value; } }

    public HybridInventory(int id, string name, double capacity, string location, string manager, string securityCode,
                           string hybridType)
        : base(id, name, capacity, location, manager, securityCode)
    {
        HybridType = hybridType;
    }

    void ITrackable.Track()
    {
        Console.WriteLine($"Отслеживание гибридного склада '{WarehouseName}'");
    }

    void IReportable.GenerateReport()
    {
        Console.WriteLine($"Отчет по гибридному складу '{WarehouseName}'");
    }

    void ISecurable.Lock()
    {
        Console.WriteLine($"Гибридный склад '{WarehouseName}' заблокирован");
    }
}

public class InventoryCollection<T> where T : Inventory
{
    private List<T> _inventories = new List<T>();
    public void Add(T inventory) => _inventories.Add(inventory);
    public int Count => _inventories.Count;
    public void DisplayAllStatuses()
    {
        foreach (var inventory in _inventories)
        {
            inventory.GetStorageStatus();
            Console.WriteLine("----------------------");
        }
    }
}

public class InventoryService
{
    private readonly Inventory _inventory;

    public InventoryService(Inventory inventory)
    {
        _inventory = inventory;
    }

    public void PerformAudit()
    {
        Console.WriteLine($"Сервис начал проверку склада '{_inventory.WarehouseName}'...");
        _inventory.Audit();
    }
}

Item item1 = new Item("Телевизор", 101, 15.5, "Электроника", "Samsung", DateTime.Now.AddYears(2));
Item item2 = new Item("Книги", 103, 5.0, "Литература", "Эксмо", DateTime.Now.AddYears(5));

PersonalInventory pers = new PersonalInventory(172, "Кладовой склад", 300.0, "Тюмень", "Амина Кадралинова", "1234", "Амина", "+7-952-123-45-67", true);
pers.GetStorageStatus();
pers.AddItem(item2);
pers.ChangeOwner("Петр", "+7-900-111-22-33");
Console.WriteLine();

GroupInventory group = new GroupInventory(212, "Склад электроники", 500.0, "Тюмень", "Иван Стариков", "5678", "Электроника", 50, "Иван");
group.GetStorageStatus();
group.AddItem(item1);
Console.WriteLine();

AutomatedInventory auto = new AutomatedInventory(385, "Строительный склад", 800.0, "Тюмень", "Сергей", "9012", 3, "R6-D7", 75, DateTime.Now.AddMonths(-2));
auto.GetStorageStatus();
auto.AddItem(item1);
auto.Charge(15);
Console.WriteLine();

SmartPersonalInventory smart = new SmartPersonalInventory(500, "Умный склад", 400.0, "Тюмень", "Петр", "3456", "Петр", "+7-952-987-65-52", true, "v2.0", false, "Алиса");
smart.GetStorageStatus();
Console.WriteLine();

HybridInventory hybrid = new HybridInventory(700, "Гибридный склад", 1000.0, "Тюмень", "Алексей", "7890", "Умный+Автоматизированный");
((ITrackable)hybrid).Track();
((IReportable)hybrid).GenerateReport();
((ISecurable)hybrid).Lock();
Console.WriteLine();

InventoryService service1 = new InventoryService(pers);
service1.PerformAudit();
Console.WriteLine();

InventoryService service2 = new InventoryService(auto);
service2.PerformAudit();
Console.WriteLine();

InventoryCollection<Inventory> allInventories = new InventoryCollection<Inventory>();
allInventories.Add(pers);
allInventories.Add(group);
allInventories.Add(auto);
allInventories.Add(smart);
allInventories.Add(hybrid);
Console.WriteLine($"Всего складов в коллекции: {allInventories.Count}");
```
